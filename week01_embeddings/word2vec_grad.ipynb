```python
import torch
```

```python
n_emb = 4
n_voc = 8

u = torch.randn(n_emb, dtype=torch.float64, requires_grad=True)
vs = torch.randn(n_emb, n_voc, dtype=torch.float64 , requires_grad=True)

v_idx = 0

v_u = vs[:, v_idx]
v = vs[:, torch.arange(n_voc) != v_idx]

J = - u @ v_u + torch.log(torch.exp(u @ v).sum(0))
J.backward()
```

```python
u_grad = -v_u + (v * torch.exp(u @ v)).sum(1) / torch.exp(u @ v).sum(0)
```

```python
torch.allclose(u_grad, u.grad)
```

```python
v_u_grad = -u
v_grad = u[:, None] * (torch.exp(u @ v)) / torch.exp(u @ v).sum(0)
vs_grad = torch.cat((
    v_u_grad[:, None], v_grad
), dim=1)
```

```python
torch.allclose(vs_grad, vs.grad)
```
